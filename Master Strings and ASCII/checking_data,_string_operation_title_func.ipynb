```python
def isalpha(text):
    c = 0
    for i in text:
        if ((ord(i) >= 65 and ord(i) <= 90) or (ord(i) >= 97 and ord(i) <= 122)):   # (65-90) Upper | (97 - 122) Lower
            c += 1
    if (len(text) == c):
        return True
    else:
        return False

print('Anuradha Singh'.isalpha())
print(isalpha('AnuradhaSingh'))
```

## **2. isdigit()**

```python
def isdigit(text):
    c = 0
    for i in text:
        if (ord(i) >= 48 and ord(i) <= 57):              # (48-57) Digits | 0-9
            c += 1

    if (len(text) == c):
        return True
    else:
        return False

print('1231'.isdigit())
print(isdigit('1231'))
```

## **3. isalnum()**

```python
def isalnum(text):
    c = 0
    for i in text:
        if ((ord(i)>=65 and ord(i)<=90) or (ord(i)>=97 and ord(i)<=122) or (ord(i)>=48 and ord(i)<=57)):
            c += 1
    if (len(text) == c):
        return True
    else:
        return False

print('AnuradhaSingh0423'.isalnum())
print(isalnum('Anu09'))
```

## **Title Func**

```python
def upper(text):
    string = ''
    for char in text:
        if ( ord(char) >= 97 and ord(char) <= 122 ):
            string += chr(ord(char) - 32)
        else:
            string += char
    return string

def lower(text):
    string = ''
    for char in text:
        if ( ord(char) >= 65 and ord(char) <= 90 ):
            string += chr(ord(char) + 32)
        else:
            string += char
    return string
```

```python
def title(text):

    output = []
    for word in text.split(' '):
        output.append(upper(word[0]) + lower(word[1:]))

    return ' '.join(output)
```

```python
title('Anuradha Singh')
```
