### **Remove Stop Words and Creating Vocab**

```python
words = []
for word in word_tokenize(corpus_1):

  if(word.lower() not in stopwords.words('english') and len(word) >= 2):
    words.append(word.lower())

vocab = list(set(words))
```

### **Creating Count Dictionary**

```python
word_freq = {}

for word in vocab:
  word_freq[word] = 0
```

### **Creating the Frequency of Words**

```python
words = []

for word in word_tokenize(corpus_1):
  if(word.lower() in stopwords.words('english')) and (len(word)>=2):
    words.append(word.lower())
```

```python
# Re-create the 'words' list that contains non-stopwords, as intended
# This addresses the issue where the previous 'words' list was incorrectly populated with stopwords.
# This ensures that we are counting words that are actual keys in 'word_freq'.
correct_words_for_counting = []
for word_token in word_tokenize(corpus_1):
  # Only include words that are not stopwords and have length >= 2
  if(word_token.lower() not in stopwords.words('english') and len(word_token) >= 2):
    correct_words_for_counting.append(word_token.lower())

# Now iterate through the correct list of words to update their frequencies
for word in correct_words_for_counting:
  # It's good practice to check if the word is in the dictionary,
  # although with 'vocab' being created from similar logic, most words should be present.
  if word in word_freq:
    word_freq[word] = word_freq[word] + 1

word_freq
```

### creating a df for sorting

```python
import pandas as pd

data = []

for key in word_freq.keys():

  data.append([key, word_freq[key]])


df = pd.DataFrame(word_freq.items(),columns=['word','count'])
print(df)
```
